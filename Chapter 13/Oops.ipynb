```python
# creat a object for a class to acces its element
class Factory:
    a=12 # attribute: variable inside class is attribute

    def hello(self): # method function inside class is method
        print("Hello Saurabh")

obj=Factory() # obj is object of Class Factory
print(obj.a)
obj.hello()
```

```python
# constructor is use to take some values from user

class factory():
    def __init__(self,name,roll,sec): # dunder method
        self.name=name #instance attribute
        self.roll=roll
        self.sec=sec

    def show(self): # instance method
        print(f"the details if rank holder is : \nName:{self.name} \nRoll: {self.roll}")

First=factory("Saurabh",1,"B")

First.show()

```

```python
#attribute and method

class Car:
    wheels=4 # class attribute

    def __init__(self, color):
        self.color=color # self.color is instance arrribute

    def speed(self):
        print("It is moving")

    @classmethod
    def stop(cls):
        print("The car is stopped")

    @staticmethod
    def engine():
        print("It has V8 engine")

C=Car("Red")

print(C.wheels)
print(C.color)
C.speed()
C.stop()
C.engine()
```
