```python
#attribute and method

class Car:
    wheels=4 # class attribute

    def __init__(self, color):
        self.color=color # self.color is instance arrribute

    def speed(self):
        print("It is moving")

    @classmethod
    def stop(cls):
        print("The car is stopped")

    @staticmethod
    def engine():
        print("It has V8 engine")

C=Car("Red")

print(C.wheels)
print(C.color)
C.speed()
C.stop()
C.engine()
```

```python
#constructor in inheritence

class Parent:
    def __init__(self,income):
        self.income=income

    def show(self):
        print(f"Parents income is: {self.income}")

class child(Parent):
    def __init__(self,income,debt):
        super().__init__(income)
        self.debt=debt
P_earn=input("Enter your Father Earning")
C_earn=input("Enter your Earning")
p=Parent(P_earn)
print(p.income)

c=child(C_earn,80000)
c.show()
print(c.income)
print(c.debt)
```

```python
#multiple inheritence 

'''
Multiple Inheritance means there will be 2 parent classes and 
only 1 child class and the child class will inherit all the 
attributes and methods of both parents
'''

class parent1:
    def __init__(self,gender):
        self.gender=gender

class parent2:
    def __init__(self,age):
        self.age=age

class child(parent1, parent2):
    def __init__(self,gender,age):
        parent1.__init__(self,gender)
        parent2.__init__(self,age)

c=child("male",45)

print(c.gender,c.age)
```

```python
#multilevel inheritence

class dadaji:
    age1=78

class pitaji(dadaji):
    age2=48

class beta(pitaji):
    age3=18


b=beta()
print(b.age3)
```

```python
class main_factory:
    def __init__(self,zips):
        self.zips=zips

class bhopal_factory(main_factory):
    def __init__(self,zips,color):
        super().__init__(zips)
        self.color=color

class pune_factory(bhopal_factory):
    def __init__(self,zips,color,pocket):
        super().__init__(zips,color)
        self.pocket=pocket

pf=pune_factory(5,"REd",4)

print(pf.zips,pf.color,pf.pocket)


```

```python
#polymorphism

class Animal:
    def sound(self):
        print("Animal makes a sound")

class dog(Animal):
    def sound(self):
        print("Woof woof")

D=dog()
A=Animal()
A.sound()
D.sound()
```

```python
#encapsulation
class Demo:
    def __init__(self):
        self.name="Saurabh Prakash"  #Public
        self._age=21                 #Protected
        self.__salary=26400          # private

D=Demo()
print(D.name)
print(D._age)
print(D.__salary) #AttributeError: 'Demo' object has no attribute '__salary'

```

```python
#abstraction

from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        print("Dog say woof")

D=Dog()
D.make_sound()
```
