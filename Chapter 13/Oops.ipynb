```python
# creat a object for a class to acces its element
class Factory:
    a=12 # attribute: variable inside class is attribute

    def hello(self): # method function inside class is method
        print("Hello Saurabh")

obj=Factory() # obj is object of Class Factory
print(obj.a)
obj.hello()
```
