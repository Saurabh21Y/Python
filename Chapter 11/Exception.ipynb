Errors:
Errors occur due to mistakes in the code that prevent it from running. These can be syntax errors or logical errors

Syntax error:
these are the type of error which occur when mw make any syntax error like,

e.g print(hello) or print("Hello"

Indentation error:
These are the error in python that shows that the indentation is not followed correctly.

e.g 
if(age>18):
print("Voter")

Exception:
Exceptions are unexpected events or errors that occurs during the execution of a program, which disrupts the normal flow of the program.


```python
inp=int(input("Enter your number: "))
try:
    div=(10/inp)
except ZeroDivisionError as e: 
# except Exception as e: will handle every error which can occur and with help of e we can access that error
    print(f"Error : {e}, you cannot input {inp}")

print("Here is your number")
```

```python
# else block: either of except or else block will be execute
inp=int(input("Enter your number: "))
try:
    div=(10/inp)

except ZeroDivisionError as e: 
# except Exception as e: will handle every error which can occur and with help of e we can access that error
    print(f"Error : {e}, you cannot input {inp}")

else:
    print(f"Here is your number {div}")

finally:
    print("This will executed")

```
