```python
#Decorator

def decorate(func):
    def wrapper():
        print("Hiiiii")
        func()
        print("How are you?")
    return wrapper



@decorate
def hello():
    print("Hello Saurabh!")

hello()
```

```python
# Args and kwargs

def add(*args):
    sum=0
    for i in args:
        sum=sum+i
    print(sum)


add(89,67,6)

def info(**kwargs):
    print("Your info is: ")

    for i in kwargs:
        print(f"{i} = {kwargs[i]}")

info(name="Saurabh", age=22, hobby="Leetcode")


```

```python
#example of decorators and args

def decorate(func):
    def wrapper(*args,**kwargs):
        print("Here we are doing the addition")
        func(*args,**kwargs)
        print("I hope this helps you")

    return wrapper


@decorate
def addition(a,b):
    print(f"The sum is {a+b}")

addition(56,78)
```
