there are two ways:
1st using index value which start from 0
2nd direct iteratoe a string using in 

```python
Name="Saurabh Prakash"
# first method
for i in range(0,len(Name)):
    print(Name[i], end=" ")
```

break ,continue, else 

break statement: if the code finds the break it will stop the execution of loop.

continue: if the code finds the continue it will skip that iteration and the continue its flow as normal.

else: this will only run with break and the condition is if the break gets executed else will not run, but if the break dont appear the else will execute for sure.


```python
#break
for i in range(1,20):
    if i==16:
        break
    else:
        print(i)

print("this is outside of loop")
```

```python
#continue
for i in range(1,20):
    if i==16:
        continue
    else:
        print(i,end=" ")
```

```python
#else:
for i in range(1,20):
    if i==45: # i<20 then break will run
        print("break found!")
        break
    print(i)

else:
    print("this is else part of loop")
```

```python
'''Accept a number and check if it a perfect number or not. 
A number whose sum of factors is equal to the number itself 
Ex -  6 = 1, 2, 3 =6
'''
num=int(input("Enter num: "))
perfect=0
for i in range(1,num):
    if num%i==0:
        perfect+=i

if(perfect==num):
    print("perfect")
else:
    print("not perfect")
```

```python
#Check wether the number is prime or not
num=int(input("Enter num: "))
prime=0
for i in range(1,num+1):
    if num%i==0:
        prime+=i

if(prime==num+1):
    print("prime")
else:
    print("not prime")
```

```python
#Reverse a string without using in build functions.
# first method.
name="Saurabh"

for i in range(-1,-(len(name))-1,-1):
    print(name[i],end="")

#second method

rev_name=""

for char in name:
    rev_name= char+rev_name

print("\n",rev_name)

```

```python
#Check string is Pallindrome or not
name=input()
rev_name=""

for char in name:
    rev_name= char+rev_name

if(name==rev_name):
    print("palindrome")
else:
    print("not palindrome")
```

```python
'''
Count all letters, digits, and special symbols from a given 
string 
Given: str1 = "P@#yn26at^&i5ve" 
Expected Outcome: 
Total counts of chars, digits, and symbols 
Chars = 8 
Digits = 3 
Symbol = 4
'''

str1 = "P@#yn26at^&i5ve"
chars=0
digits=0
symbol=0
for i in str1:
    if i.isdigit():
        digits+=1
    elif i.isalpha():
        chars+=1
    else:
        symbol+=1

print(chars, digits, symbol)
```
