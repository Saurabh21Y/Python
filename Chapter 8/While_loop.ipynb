```python
#Accept a number and print its reverse
num=int(input("enter the number"))
rem=0
rev=0
while(num>0):
    rem=num%10
    rev=rem+(rev*10)
    num=num//10

print(rev)
```

```python
#Accept a number and check if it is a pallindromic number
num=int(input("enter the number"))
safe=num
rem=0
rev=0
while(num>0):
    rem=num%10
    rev=rem+(rev*10)
    num=num//10

if(rev==safe):
    print("Its palindrome")
else:
    print("non palindrome")
```

```python
#Create a random number guessing game with python
import random
num=random.randint(-999,999)
guess=int(input("Guess the number: "))
while(num!=guess):
    if(guess<num):
        print("the value is greater")
        guess=int(input("Guess the number again: "))
    elif(guess>num):
        print("the value is lesser")
        guess=int(input("Guess the number again: "))

print("You won!")
```
