```python
# set is unordered meand we cannot acces it through index value.
sett={1,2,3,4,5}
print(sett[0]) # will throw error
```

Set is semi-heterogenous it can store some data types like string, numbers, tuples but not everything

speciaaly thode datatype who is ordered will not be aaceptable to store in set.

```python
# set methods to use mutable behaviour of 

sett={1,2,3,4,5}

sett.add(6) # this will add a element to the set
print(sett)

sett.remove(6) # Removes 6 (It can raise error if does not find the element 6)
print(sett)

sett.discard(6) # Removes 6 (It will not raise error if does not find the element 6)
print(sett)

popped=sett.pop() # this will pop a random element and return it
print(popped)
print(sett)

sett.clear() # this will remove all element from the set
print(sett)
```

```python
# some special set methods
a={1,2,3}
b={3,4,5}

union_set=a.union(b) # equivalent to (a|b)
print(union_set)

intersection=a.intersection(b) # equivalent to (a&b)
print(intersection)

diff_set=a.difference(b) # equivalent to (a-b)
print(diff_set)

sym_diff=a.symmetric_difference(b) # equivalent to (a^b)
print(sym_diff)
```
