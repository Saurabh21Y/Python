```python
# Initial list
numbers = [5, 2, 9, 1, 5, 6]  
print("Initial list:                    ", numbers)

# 1. append() - Adds an element to the end
numbers.append(10)  
print("After append(10):                ", numbers)

# 2. insert() - Inserts element at a specific index
numbers.insert(2, 15)  
print("After insert(2, 15):             ", numbers)

# 3. extend() - Appends elements from another list to the end
numbers.extend([20, 25, 30])  
print("After extend([20, 25, 30]):      ", numbers)

# 4. remove() - Removes the first occurrence of a value
numbers.remove(5)  
print("After remove(5):                 ", numbers)

# 5. pop() - Removes and returns the element at a specific index
popped_item = numbers.pop(3)  
print(f"Popped item at index 3:           {popped_item}")
print("After pop(3):                    ", numbers)

# 6. index() - Returns the index of the first occurrence of a value
idx_of_6 = numbers.index(6)  
print("Index of 6:                      ", idx_of_6)

# 7. count() - Returns the number of times a value appears
count_5 = numbers.count(5)  
print("Count of 5:                      ", count_5)

# 8. sort() - Sorts the list in ascending order (in-place)
numbers.sort()  
print("After sort():                    ", numbers)

# 9. reverse() - Reverses the order of the list (in-place)
numbers.reverse()  
print("After reverse():                 ", numbers)

# 10. copy() - Returns a shallow copy of the list
new_numbers = numbers.copy()  
print("Copied list (new_numbers):       ", new_numbers)

# 11. clear() - Removes all elements from the list
numbers.clear()  
print("After clear():                   ", numbers)
```

```python
# Print positive and negative elements of an List
lst=[1,-3,6,-4,-6,8,9,3,7,56,-89]
pos=[]
neg=[]

def checker(n):
    if n >= 0 :
        return True
    else:
        return False

for i in lst:
    if checker(i) :
        pos.append(i)
    else:
        neg.append(i)

print(f"Positive: {pos}")
print(f"Negative: {neg}")
```

```python
# Mean of List elements
lst=[1,3,6,4,6,8,9,3,7,56,89,78]

sum=0

for i in lst:
    sum=sum+i

print(sum/len(lst))

```

```python
#Find the greatest element and print its index too

lst = [12, 45, 2, 89, 34, 89, 7,90]

max_val=lst[0]
max_idx=0

for i in range(1,len(lst)):
    if lst[i]>max_val:
        max_val=lst[i]
        max_idx=i

print(f"The maximum value is {max_val} and its index is {max_idx}")

```

```python
#Find the second greatest element

lst = [12, 45, 2, 89, 34, 87, 7,90]

max=max2=lst[0]

for i in range(1,len(lst)):
    if lst[i]>max:
        max2=max
        max=lst[i]

print(max2)

```

```python
#Check if List is sorted or not. 

lst=[1,2,3,4,5,6,7,8,9,9,9,10,34,23]

sorted=True

for i in range(len(lst)-1):
    if lst[i+1]>=lst[i]:
        sorted=True
    else:
        sorted=False

print(sorted)
```
