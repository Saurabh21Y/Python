```python
# mutable: we can only chnage the value 

d={
    1:"Rohan",
    2:"Rahul",
    3:"Saurabh",
    4:"Deva",
    5:"Rohan",

}

d[3]="Admin"

print(d)
```

```python
# traversing over dict

d={
    1:"Rohan",
    2:"Rahul",
    3:"Saurabh",
    4:"Deva",
    5:"Rohan",
}

for i in d: #d.values() will iterate the values of the dict
    print(f"{i}: {d[i]}")
```

```python
#Write a Python program to sum all the values in a dictionary
d1={1:10,2:20,3:30,4:50}
sum=0
for i in d1:
    sum=sum+d1[i]

print(sum)

```

```python
#Count the frequency of each element of the list

lst=[1,1,1,2,2,3,2,4,5,4,3,5,6,7,7,6,7,8,8,7,8,7,5,4,3,4]
d={}
for i in lst:
    if i in d.keys():
        d[i] +=1
    else:
        d[i]=1

print(d)
    
```
