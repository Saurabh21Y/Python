```python
# make a tuple
tup=(1,2,3,4,5)
print(type(tup))
```

```python
# tuple is immutable
tup=(1,2,3,4,5)
print(tup[0]) # this is possible because tuple is ordered we can access value through index 
# tup[0]=7 # this will throw a error
print(tup[0])
```

```python
# tuple can store duplicate value
tup=(1,1,2,2,3,3,"s",print(),True)
print(tup)
```

```python
# traversing
tup=(1,2,3,4,5,6)

#first method
for i in tup:
    print(i)

# second method
for i in range(len(tup)):
    print(tup[i])
```

```python
# tuple unpacking

a,b,c,d=(1,2,3,4)
print(a)
```
